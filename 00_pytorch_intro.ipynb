###manipulating tensors
#operations  involve additions, multiplications-elementwise and matrix multiplication, modifications, divisions and substractions

```python
tensor1 = torch.tensor([1,2,3])
tensor2 = torch.tensor([4,5,6])
constant_add = tensor1+10
print(constant_add)
added_tensor = tensor1+tensor2
print(added_tensor)
# multiplied_tensor = tensor1*tensor2
# multiplied_tensor = tensor1.mul(tensor2)
multiplied_tensor=torch.matmul(tensor1, tensor2)
print(multiplied_tensor)
substracted_tensor = tensor1-tensor2
print(substracted_tensor)
```

##Multiplicaitons -> constant multiplications and matrix multiplications vector multiplications and multi-dimentional matrix multiplications


## methods in pytorch: torch.mm, torch.matmul, torch.bmm

#####important to note, element-wise multiplicaiotn is different than matrix multiplication(can also refer as dot product)

```python
tensor1 = torch.rand(2,3,4)
tensor2 = torch.rand(2,4,6)

resultant = torch.bmm(tensor1,tensor2)
print(resultant.shape)
resultant

#result dimensions will always be outer dimensions of both matrix eg, MxN * NXB = MxB
```

```python
#transpose

t1 = torch.rand(3,4)
t2 = torch.rand(3,4)
res = torch.mm(t1,t2.T)
res
```

```python
#reshaping, stacking, squeezing and unsqueezing
#reshaping: reshapes an input tensor to a defined shape
#view: return a view of an input tensor of certain shape but keep the memory same,(same tensor different shape)
#stackin:combining multiple tensors on to one, stack, vstack, hstack
#squeeze: removes all '1' dimensions from the tensor
#unsqueeze: add '1' dimension to the target tensor
#permute: return a view of the input with the dimensions permuted(swapped) in certain way.


tensor = torch.arange(1.,10.)
# tensor, tensor.shape
reshaped_tensor = tensor.reshape(1,3,3)
tensor.shape, reshaped_tensor, reshaped_tensor.shape
```

```python
#change the view: view creates the different metadata but points to the same data so change in a view data =>change in original

z = tensor.view(1,9)
z
```

```python
x = torch.randint(10,(3,3))
y = torch.randint(10,(3,3))
x,y,x.shape, y.shape
```

```python
z = torch.stack((x,y),dim = 1)
z, z.shape

```

```python
z = torch.stack((x,y),dim = 2)
z, z.shape
```

```python
#squeeze and unsqueeze

# squeeze: removes all single dimensions from the tensor and unsqueeze: adds

z.squeeze()
```

```python
z.unsqueeze(dim=3)
```
