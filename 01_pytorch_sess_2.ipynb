##usually startoff with data in numpy array and you want it in tensor

##there are ways to convert the same in torch and also for vice-versa

```python
np_arr = np.arange(1,8, dtype=np.float32)
# np uses dtype 64 by default
tensor = torch.from_numpy(np_arr)
print(np_arr)
np_arr+=1 #memory sharing
print(np_arr)
print(tensor)
#
#changing the np_arr will result in modification of the tensor if we use from_numpy
# if we use torch.tensor(np_arr) -->result in separate memory allocated for the tensor, and no memory sharing
```

```python
tensor_to_np = tensor.numpy()
# tensor_to_np
tensor.add_(1)
tensor_to_np,tensor
```

###Reproducability

##in short the nn starts with random numbers->perform calculations-->update the random numbers and try to make better representaion of the data->again -> again

#to reduce the randomness in nn, the pytorch has a concept of randomseed

##essentially what the randomseed does is to flavour the randomness

```python
tensor_rand_a = torch.rand(3,4)
tensor_rand_b = torch.rand(3,4)
print(tensor_rand_a,tensor_rand_b)
print(torch.equal(tensor_rand_a,tensor_rand_b))
```

```python
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)

tensor_rand_a = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
tensor_rand_b = torch.rand(3,4)

print(tensor_rand_a==tensor_rand_b)
```
